# ORNL_LOTF container check
Open with the **ORNL_LOTF (container)** kernel. Confirms the kernel runs inside the image, that the pinned packages import, and that HeteroBuilder's py3Dmol viewer renders.

```python
import os, sys, importlib.metadata as md
print(sys.executable, sys.version.split()[0])
print('in container:', os.path.exists('/opt/ornl_lotf_pip_freeze.txt'))
for pkg in ['vdW_structures', 'EnsembleFFFit', 'matensemble', 'pymatgen', 'mp-api', 'papermill', 'py3Dmol']:
    print(f'{pkg:16s}', md.version(pkg))
print('MP_API_KEY set:', bool(os.environ.get('MP_API_KEY')))
```

```python
from pymatgen.core import Lattice, Structure
from vdW_structures.vdW_structure import VdWStructure
from vdW_structures.viewer import view_structure

a, c = 3.8, 24.0
lat = Lattice([[a, 0, 0], [-a / 2, a * 3**0.5 / 2, 0], [0, 0, c]])
z = [0.10, 0.17, 0.035, 0.60, 0.67, 0.535]
coords = [[0, 0, z[0]], [1/3, 2/3, z[1]], [2/3, 1/3, z[2]], [0, 0, z[3]], [1/3, 2/3, z[4]], [2/3, 1/3, z[5]]]
toy = Structure(lat, ['Cr', 'Te', 'Te'] * 2, coords).make_supercell([3, 3, 1])
vdw = VdWStructure(toy, minimum_vdW_gap=2.5)
print('layers:', len(vdw.vdW_layers), 'spacings:', vdw.vdW_spacings)
view_structure(vdw.structure, {'Cr': 'blue', 'Te': 'orange'})
```
